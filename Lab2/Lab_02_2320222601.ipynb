## S03 Supplied preprocessing and query helpers

<div dir="rtl" lang="ar">شغل دون تعديل. BASELINE يحدد إعدادات المعالجة. تبدأ المواقع من الصفر داخل كل حقل. record_terms يدمج كلمات العنوان والوصف ووسوم الكتالوج للفهرس المعكوس، أما الفهرس الموضعي فيحفظ كلمات title وsummary فقط. لا يوجد تجذير أو ترجمة.</div>

This tokenizer improves on Lab 01 by retaining digits and attached combining marks. The helper `all_terms_query` accepts plain query words with implicit AND, not a Boolean expression parser. Never pass `topic_python` to this text helper: look up tag postings directly.

```python
# S03 Run supplied helpers
"""Supplied scaffolding for Lab 02. The core algorithms are student exercises."""
import re
import unicodedata

BASELINE = {'casefold': True, 'arabic_fold': False, 'remove_stopwords': False}
STOP_WORDS = {'a','an','the','and','or','to','from','in','of','for','with','في','من','على','و'}
FIELDS = ('title', 'summary')

def normalize_text(text, config):
    text = unicodedata.normalize('NFKC', text)
    if config['arabic_fold']:
        text = re.sub(r'[\u064b-\u065f\u0670\u0640]', '', text)
        text = text.translate(str.maketrans({'أ':'ا','إ':'ا','آ':'ا','ٱ':'ا','ى':'ي'}))
    return text.casefold() if config['casefold'] else text

def tokenize_with_positions(text, config):
    # Keep attached combining marks inside letter/digit runs.
    tokens, current = [], []
    for char in normalize_text(text, config):
        if char.isalnum() or (unicodedata.category(char).startswith('M') and current):
            current.append(char)
        elif current:
            tokens.append(''.join(current)); current = []
    if current:
        tokens.append(''.join(current))
    # Preserve original token offsets if the experimental stop list removes words.
    return [(term, position) for position, term in enumerate(tokens)
            if not config['remove_stopwords'] or term not in STOP_WORDS]

def preprocess(text, config):
    return [term for term, _ in tokenize_with_positions(text, config)]

def catalog_tags(record):
    return {'topic_' + topic for topic in record['topics']} | {
        'lang_' + record['language'], 'level_' + record['level'],
        'access_' + record['access'], 'type_' + record['format']}

def record_terms(record, config):
    # Keep structured catalog tags atomic. They are not automatically translated.
    terms = set(catalog_tags(record))
    for field in FIELDS:
        terms.update(preprocess(record[field], config))
    return terms

def union_sorted(a, b):
    """Supplied OR helper. The AND merge is a student exercise."""
    return sorted(set(a) | set(b))

def complement_sorted(postings, universe):
    """NOT applies only to the current collection."""
    return sorted(set(universe) - set(postings))

def all_terms_query(index, query_text, config, intersect):
    terms = preprocess(query_text, config)
    if not terms:
        return []
    result = index.get(terms[0], [])
    for term in terms[1:]:
        result = intersect(result, index.get(term, []))
    return result

def field_terms_query(index, field, query_text, config):
    """Co-occurrence within one field, not necessarily an exact phrase."""
    terms = preprocess(query_text, config)
    if not terms:
        return []
    matches = [{doc for doc, fields in index.get(term, {}).items() if field in fields}
               for term in terms]
    return sorted(set.intersection(*matches))

def phrase_query(index, query_text, config, matcher):
    if config['remove_stopwords']:
        raise ValueError('Exact phrase search in this lab keeps stop words')
    return matcher(index, preprocess(query_text, config))

def near_matches(index, first, second, k=3):
    """Optional unordered proximity. abs(position1-position2) <= k, same field."""
    if k < 0:
        raise ValueError('Distance must be nonnegative')
    result = []
    for doc in sorted(set(index.get(first, {})) & set(index.get(second, {}))):
        common_fields = set(index[first][doc]) & set(index[second][doc])
        if any(abs(a-b) <= k and (first != second or a != b)
               for field in common_fields
               for a in index[first][doc][field]
               for b in index[second][doc][field]):
            result.append(doc)
    return result

config = dict(BASELINE)
print("Baseline:", config)
print("Title token positions:", tokenize_with_positions("Introduction to Information Retrieval", config))
```

```python
# S05 Your A and B functions
def build_inverted_index(records, config):
   
    # create a dictionry to hold the inverted index
    index = {}

    # loop through each record in the records list
    for record in records:
        doc_id = record['id']
        terms = record_terms(record, config)

        # Posting Lists
        for term in terms:
            if term not in index:
                index[term] = set()

            index[term].add(doc_id)

    # sort the posting lists for each term in the index
    for term in index:
        index[term] = sorted(index[term])

    return index

    # raise NotImplementedError("Complete A build_inverted_index")

def intersect_sorted(a, b):
    # AND (Intersection)
    i = 0
    j = 0
    result = []

    while i < len(a) and j < len(b):
        if a[i] == b[j]:
            result.append(a[i])
            i += 1
            j += 1

        elif a[i] < b[j]:
            i += 1

        else:
            j += 1

    return result

    # raise NotImplementedError("Complete B intersect_sorted")
```

## S08 Complete C and D

<div dir="rtl" lang="ar">أكمل الفهرس الموضعي ومطابقة العبارات وفق الصفحتين 9 و10. احتفظ بجميع التكرارات وافصل title وsummary. لا تدخل وسوم الكتالوج في المواقع. فهرس العبارات الأساسي يرفض إزالة كلمات الوقف.</div>

A positional entry has the structure `index[term][doc_id][field] = [positions]`. Query terms remain an ordered list, including repeats.

```python
# S08 Your C and D functions
def build_positional_index(records, config):
    # C: reject remove_stopwords=True with ValueError for this phrase baseline.
    # Process title and summary separately using tokenize_with_positions.
    # Return {term: {doc_id: {field: [all occurrence positions]}}}.

    if config['remove_stopwords']:
        raise ValueError("Stop-word removal is not allowed")

    index = {}

    for record in records:
        doc_id = record['id']

        for field in FIELDS:
            tokens = tokenize_with_positions(
                record[field], config
            )

            for term, position in tokens:

                if term not in index:
                    index[term] = {}

                if doc_id not in index[term]:
                    index[term][doc_id] = {}

                if field not in index[term][doc_id]:
                    index[term][doc_id][field] = []

                index[term][doc_id][field].append(position)

    return index

    # raise NotImplementedError("Complete C build_positional_index")

def phrase_matches(index, query_terms):
    # D: query_terms is an already-preprocessed list of terms.
    # Empty list or unknown terms -> [].
    # Require start + offset for every term, within ONE field.
    # Preserve repeated query terms; do not convert the query to a set.
    # Return each matching document once, with IDs sorted.
    # raise NotImplementedError("Complete D phrase_matches")
    
    if not query_terms:
        return []

    for term in query_terms:
        if term not in index:
            return []

    candidates = set(index[query_terms[0]])

    for term in query_terms[1:]:
        candidates = candidates.intersection(index[term])

    result = set()

    for doc_id in candidates:

        first_term = query_terms[0]

        for field in index[first_term][doc_id]:

            for start in index[first_term][doc_id][field]:

                matched = True

                for offset, term in enumerate(query_terms):

                    positions = index[term][doc_id].get(field, [])

                    if start + offset not in positions:
                        matched = False
                        break

                if matched:
                    result.add(doc_id)

    return sorted(result)

```

**B:** Trace every pointer comparison for [R01,R03,R05,R09] and [R02,R03,R09,R12]. Explain O(m+n) and why moving the smaller pointer is safe.


**Two-Pointer Intersection**

Input lists:

`a = ['R01', 'R03', 'R05', 'R09']`

`b = ['R02', 'R03', 'R09', 'R12']`

| Step | i | j | a[i] | b[j] | Action | Result |
|---|---|---|---|---|---|---|
| 1 | 0 | 0 | R01 | R02 | Move i | [] |
| 2 | 1 | 0 | R03 | R02 | Move j | [] |
| 3 | 1 | 1 | R03 | R03 | Add R03, move both | [R03] |
| 4 | 2 | 2 | R05 | R09 | Move i | [R03] |
| 5 | 3 | 2 | R09 | R09 | Add R09, move both | [R03, R09] |

**Final result:** `['R03', 'R09']`

After the last comparison, `i = 4` and `j = 3`. The algorithm stops because the first list has ended.

Moving the pointer with the smaller ID is safe because both posting lists are sorted. The smaller ID cannot match any later ID in the other list.

The time complexity is **O(m + n)**, where m and n are the lengths of the two posting lists. Each pointer moves forward only and never goes backward.
